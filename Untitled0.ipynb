```python
import pandas as pd

URL_BASE = "https://raw.githubusercontent.com/lucioiacovetta/Pre-entrega-semana-2/main/"

ventas = pd.read_csv(URL_BASE + "transacciones.csv")
productos = pd.read_excel(URL_BASE + "maestro_productos.xlsx", engine="openpyxl")

print(ventas.shape, productos.shape)
print(ventas.dtypes)
print(productos.dtypes)
ventas.head()

















```

```python
ventas = pd.read_csv(URL_BASE + "transacciones.csv",
                     dtype={"id_factura": str, "id_producto": str},
                     low_memory=False)

# Elimino las filas completamente vacías (quedaron de recortar el archivo en Excel)
print("Filas antes:", len(ventas))
ventas = ventas.dropna(how="all")
print("Filas después:", len(ventas))

# Fecha: formato día-mes-año (2 dígitos) hora:minuto
ventas["fecha"] = pd.to_datetime(ventas["fecha"], format="%d-%m-%y %H:%M", errors="coerce")

print(ventas.dtypes)
print("Fechas que no se pudieron convertir:", ventas["fecha"].isna().sum())
print("Rango:", ventas["fecha"].min(), "→", ventas["fecha"].max())
```

```python
# --- Terminar de corregir tipos ---
ventas["cantidad"] = ventas["cantidad"].astype(int)
ventas["id_cliente"] = ventas["id_cliente"].astype("Int64")

# Llave en el mismo tipo y formato en ambas tablas
ventas["id_producto"] = ventas["id_producto"].astype(str).str.strip().str.upper()
productos["id_producto"] = productos["id_producto"].astype(str).str.strip().str.upper()

print(ventas.dtypes)
print(productos.dtypes)

# --- 3.4 Valores nulos: primero los cuento ---
print("\nNulos en ventas:\n", ventas.isna().sum())
print("\nNulos en productos:\n", productos.isna().sum())
```

```python
# --- 3.4 Manejo de nulos ---

# Ventas: imputo cliente faltante
ventas["id_cliente"] = ventas["id_cliente"].fillna(0)

# Productos: elimino los que no tienen precio, imputo nombre faltante
productos = productos.dropna(subset=["precio_unitario"])
productos["nombre_producto"] = productos["nombre_producto"].fillna("SIN DESCRIPCION")

# Duplicados en la llave del maestro
print("IDs duplicados en productos:", productos["id_producto"].duplicated().sum())
productos = productos.drop_duplicates(subset="id_producto", keep="first")

print("\nNulos en ventas:\n", ventas.isna().sum())
print("\nNulos en productos:\n", productos.isna().sum())
print("\nProductos que quedan:", len(productos))
```

```python
# --- 3.5 Merge por id_producto ---
df = ventas.merge(
    productos,
    on="id_producto",
    how="left",
    validate="many_to_one",
    indicator=True
)

print(df["_merge"].value_counts())
print("Filas ventas:", len(ventas), "| Filas después del merge:", len(df))
```

```python
# Elimino las ventas sin producto en el maestro y la columna auxiliar
df = df[df["_merge"] == "both"].drop(columns="_merge")

# --- 4. Columna total_venta ---
df["total_venta"] = df["cantidad"] * df["precio_unitario"]

print(df.shape)
print(df[["id_producto", "nombre_producto", "cantidad", "precio_unitario", "total_venta"]].head())
print("\nVentas con total negativo (devoluciones):", (df["total_venta"] < 0).sum())
print("Total vendido:", round(df["total_venta"].sum(), 2))
```

```python
# --- 5. Exportación ---
df.to_parquet("ventas_integradas.parquet", index=False)

# Verifico que se lea bien y que conserve los tipos
check = pd.read_parquet("ventas_integradas.parquet")
print(check.dtypes)
print(check.shape)

# Descargo el archivo para subirlo al repo
from google.colab import files
files.download("ventas_integradas.parquet")
```
